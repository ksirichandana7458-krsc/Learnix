```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/sample_data/StudentsMarks.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
```

```python
averagemarks = df['Marks'].mean()
print('AverageMarks :', round(averagemarks, 2))
```

```python
plt.figure(figsize=(10,5))
plt.bar(df['Name'], df['Marks'], color='skyblue')
plt.title('Students Marks')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
